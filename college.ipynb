```python
data={
    "Age":[25,30,np.nan,35,22],
    "Height":[6,np.nan,5,np.nan,6],
    "Grade":['A','B','C','A',np.nan]
}
```

```python
dff=pd.DataFrame(data)
```

```python
before=dff.copy()
```

```python
dff.isnull().sum()
```

```python
dff['Age']=dff['Age'].fillna(dff['Age'].mean())
```

```python
dff['Height']=dff['Height'].fillna(dff['Height'].median())
```

```python
dff=dff.dropna()
```

```python
dff.isnull().sum()
```

```python
print("BEFORE:")
display(before)

print("AFTER:")
display(dff)
```
